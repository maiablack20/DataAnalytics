```python
print(r_4)
shuffle_1 = np.random.shuffle(r_4)
print(r_4)
```

### What I found when using np.random.shuffle is that the the rows in the second array are now the oppposite of the first array, which goes from smallest to largest. The second array goes from largest to smallest in the sequence, but none of the variables in the arrays are any different. 

```python
new_shuffle = np.random.permutation(r_4)
print(new_shuffle)
print("===========================================================")
print(r_4)
```

### The difference that I found between shuffle and permutation is that permutation randomizes the ordering of the numerical values while shuffle just shuffles around the order. Each one does it at random, but permutation really makes it become even more random.

## Grocery Problem

```python
bag1 = [2, 3, 4, 0]
bag2 = [4, 0, 1, 2]

# for bag1: 2 = vegetables, 3 = fruits, 4 = boxes of snacks, 0 = no beverages
# for bag2: 4 = vegetables, 0 = no fruits, 1 = box of snacks, 2 = beverages
```

```python
gb1 = np.array(bag1)
print(gb1)

print("===================================================")

gb2 = np.array(bag2)
print(gb2)
```

```python
grocery_a = np.sum(gb1)
grocery_b = np.sum(gb2)

print("===================================================")

print(f"The first bag contains {grocery_a} items, and the second bag contains {grocery_b} items.")
```

```python
grocery_c = np.add(gb1, gb2)
print(f"There are {grocery_c[0]} vegetables, {grocery_c[1]} fruits, {grocery_c[2]} boxes of snacks, and {grocery_c[3]} beverages in two grocery bags.")
```

## Teamwork

```python
Ali = [5, 4, 6, 4, 2]
Ben = [0, 5, 5, 3, 4]
Carla = [3, 6, 0, 3, 4]
```

```python
Ali_1 = np.array(Ali)
print(Ali_1)

print("=====================================================")

Ben_1 = np.array(Ben)
print(Ben_1)

print("======================================================")

Carla_1 = np.array(Carla)
print(Carla_1)
```

```python
all_knocks = np.vstack([Ali_1, Ben_1, Carla_1])
print(all_knocks)
```

```python
knocks_1 = np.sum(all_knocks, axis = 1)
print(knocks_1)
```

```python
knocks_2 = np.sum(all_knocks)
print(knocks_2)
```

```python
knocks_3 = np.cumsum(all_knocks, axis = 0)/2
print(knocks_3)
```
