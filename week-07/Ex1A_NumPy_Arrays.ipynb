# Looping Through Arrays

```python
numbers = [(11, 12, 13),
        (14, 15, 16)]
arrA = np.array(numbers)
print(arrA)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

### The noticeable difference between using for x in arrA with a print(x * 2) is that it prints two columns of variables that have been the result of the originals being multiplied by 2 and the difference in the print(y * 2) is that it gives you one singular line of variables multiplied by 2.

```python
for x in arrA: 
    for y in x: 
        loop_result = x * 2
        print(loop_result)
```

## This result is different from the first loop with y because it replicates the results three different times for 22, 24, 26, 28, 30, and 32 rather than just giving us the variables once. It prints three times because it's in a loop and there isn't anything to stop it. 

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate([arrA, arrA], axis = 0)
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate([arrA, arrA], axis = 1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack([joined_array_1, joined_array_1], axis = 0)
print(stacked_array_1)
```

```python
stacked_array_2 = np.stack([joined_array_1, joined_array_2], axis = 0)
print(stacked_array_2)

# I was given a ValueError that stated all input arrays must have the same shape. 
```

```python
another_joined_array = np.concatenate([joined_array_1, arrA])
print(another_joined_array)
```

```python
another_2 = np.concatenate([joined_array_1, arrA], axis = 1)
print(another_2)

# The error received was a ValueError, which stated that the dimensions of the arrays must be an exact match and this excludes the axis. 
```

```python
stacked_1 = np.hstack([arrA, arrA])
print(stacked_1)
```

```python
# What the hstack function does is repeat the variables within each square bracket, which is just two, and then stacks them horizontally on top of each other. 
```

```python
stacked_2 = np.vstack([arrA, arrA])
print(stacked_2)
```

```python
# With the vstack function, it creates duplicates of the array then it stacks them vertically on top of each other. 
```

```python
stacked_3 = np.dstack([arrA, arrA])
print(stacked_3)
```

```python
# Finally, with the dstack function, it separates the duplicates of the variables, such as 11, 12, and 13, then it stacks it on top of the duplicates as well, ordering them from smallest to largest. 
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')
print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = [(7, 6), 
        (13, 5),
        (7, 3)]
arrB = np.array(arrB)
arrB_sorted = np.sort(arrB)
print(arrB_sorted)
```

```python
print(arrB)
z = arrB < 7
arrC = arrB[z]
print(arrC)
```

```python
print(arrC)
y = arrB >= 7 
arrD = arrB[y]
print(arrD)
```
