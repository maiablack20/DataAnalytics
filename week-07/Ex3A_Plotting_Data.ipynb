# Exercise 3.A Getting familiar with Matplotlib

```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 7, 11]
```

```python
plt.plot(x, y)
plt.show()
```

```python
plt.plot(x, y, marker = 'o')
plt.show()
```

```python
plt.plot(x, y, marker = 'o')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

## Styling Plots

```python
plt.plot(x, y, linestyle = '--', color = 'red')
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

```python
plt.plot(x, y, linestyle = '--', color = 'green', marker = 's', markersize = 10)
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

```python
plt.plot(x, y, linestyle = '--', color = 'pink', marker = 'D', markersize = 12)
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

```python
plt.plot(x, y, linestyle = '--', color = 'yellow', marker = '*', markersize = 23)
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.show()
```

# Exercise 3.B More plotting practice

## Advanced Plotting Techniques

```python
products = ['Product A', 'Product B', 'Product C', 'Product D']
sales = [23, 17, 35, 29]
```

```python
plt.bar(products, sales, color = 'skyblue', width = 0.5)
plt.title('Sales By Product')
plt.xlabel('products')
plt.ylabel('sales')
plt.show()
```

```python
x = np.random.rand(50)
y = np.random.rand(50)
s = np.random.rand(50) * 100
```

```python
plt.scatter(x, y, s, cmap = 'viridis', c = s)
plt.title('SkyBlue Chart')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

## Subplots

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
```

```python
fig, axs = plt.subplots(1, 2)
axs[0].plot(x, y)
axs[1].hist(np.random.randn(1000))
axs[0].set_title('Sine Wave')
axs[1].set_title('Histogram')
axs[0].set_xlabel('X-axis')
axs[0].set_ylabel('Y-axis')
axs[1].set_xlabel('X-axis')
axs[1].set_ylabel('Y-axis')
```
