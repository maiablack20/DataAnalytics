# Exercise 2.C Working with multiple DataFrames

```python
import numpy as np
import pandas as pd
```

## Basic Aggregations

```python
data = {'Product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'],
            'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North'],
            'Units_Sold': [10, 15, 20, 5, 10, 15, 10, 5, 15],
            'Price': [100, 200, 150, 100, 200, 150, 100, 200, 150]}
```

```python
df_sales = pd.DataFrame(data)

df_sales['Revenue'] = df_sales['Units_Sold'] * df_sales['Price']
df_sales
```

```python
total_units = sum(df_sales['Units_Sold'])
print(total_units)
```

```python
total_rev = sum(df_sales['Revenue'])
print(total_rev)
```

```python
avg_units = df_sales['Units_Sold'].mean()
print(avg_units)
```

```python
avg_rev = df_sales['Revenue'].mean()
print(avg_rev)
```

## Grouping Data

```python
grpdta_1 = df_sales.groupby('Product')
print(grpdta_1)
```

```python
sum_units = grpdta_1[['Units_Sold', 'Revenue']].sum()
print(sum_units)
```

```python
grpdta_2 = df_sales.groupby('Region')
print(grpdta_2)
```

```python
mean_u_region = grpdta_2[['Units_Sold', 'Revenue']].mean()
print(mean_u_region)
```

```python
grpdta_3 = df_sales.groupby(['Product', 'Region'])
print(grpdta_3)
```

```python
sum_r_u = grpdta_3[['Units_Sold', 'Revenue']].sum()
print(sum_r_u)
```

## Sorting and Ranking Data

```python
sorted_1 = df_sales.sort_values(['Units_Sold'], ascending = True)
print(sorted_1)
```

```python
sorted_2 = df_sales.sort_values(['Revenue'], ascending = False)
print(sorted_2)
```

```python
df_sales['Revenue_Rank'] = df_sales['Revenue'].rank(ascending = False)
df_sales
```

## Merging DataFrames

```python
product_data = {'Product': ['A', 'B', 'C'],
                'Category': ['Electronics', 'Furniture', 'Appliances']}
df_products = pd.DataFrame(product_data)
df_products
```

```python
df_merged = df_sales.merge(df_products)
df_merged
```

```python
total_rev1 = df_merged.groupby('Category')['Revenue'].sum()
print(total_rev1)
```
