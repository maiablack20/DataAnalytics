### This notebook analyzes a small sales dataset using Python functions. Each function takes the full dataset as its first argument. 

```python
sales_data = [('Laptop', 'East', 1200.00),
              ('Phone', 'West', 850.00),
              ('Tablet', 'East', 400.00),
              ('Monitor', 'South', 650.00),
              ('Keyboard', 'West', 1150.00),
              ('Phone', 'East', 900.00),
              ('Tablet', 'West', 425.00),
              ('Monitor', 'East', 680.00),
              ('Keyboard', 'South', 80.00), 
              ('Laptop', 'West', 1300.00),
              ('Phone', 'West', 820.00)]
```

## Lab 1. Total sales by region

```python
def region_total(data, region):
    return sum(amt for _, r, amt in data if r == region)
```

```python
print(region_total(sales_data, 'East'))
print(region_total(sales_data, 'West'))
print(region_total(sales_data, 'South'))
```

## Lab 2. Count of sales by region

```python
def region_count(data, region):
    return sum(1 for _, r, _ in data if r == region)
```

```python
print(region_count(sales_data, 'East'))
print(region_count(sales_data, 'South'))
print(region_count(sales_data, 'West'))
```

## Lab 3. Average sale amount by region

```python
def region_average(data, region):
    if region_count(data, region) == 0:
        return 0 
    avg_region = region_total(data, region)/region_count(data, region)
    return avg_region
```

```python
print(region_average(sales_data, 'East'))
print(region_average(sales_data, 'South'))
print(region_average(sales_data, 'West'))
print(region_average(sales_data, 'North'))
```

## Lab 4. Filter by sale amount

```python
def sales_above(data, threshold):
    sales_thr = []
    for i in data: 
        if i[2] > threshold: 
            sales_thr.append(i)
    return sales_thr
```

```python
for i in sales_above(sales_data, 500):
    print(i) 
        
```

```python
def sales_below(data, threshold):
    sales_thr = []
    for i in data: 
        if i[2] < threshold: 
            sales_thr.append(i)
    return sales_thr
```

```python
for i in sales_below(sales_data, 500): 
    print(i)
```

## Lab 5. Top and bottom sales

```python
def top_sale(data):
    return max(data, key=lambda s: s[2])
```

```python
def bottom_sale(data):
    return min(data, key=lambda s: s[2])
```

```python
print(top_sale(sales_data))
print(bottom_sale(sales_data))
```

# The top sale was a Laptop in the West region for $ 1300.00

# The bottom sale was a Keyboard in the South region for $80.00

# Lab 6. Apply a discount

```python
def apply_discount(data, discount_pct):
    return list(map(lambda s: (s[0], s[1], s[2] * (1 - discount_pct)), data))
```

```python
print((apply_discount(sales_data, 0.10)))
```

```python
print((apply_discount(sales_data, 0.15)))
```

# Lab 7. Unique regions

```python
def unique_regions(data):
    return sorted(set(s[1] for s in data))
```

```python
print(sales_data)
```

I feel this function helps to keep things a little more organized and separate for the analyst, which makes it easier to filter through and observe. 

## Lab 8. Region summary

```python
def region_summary(data, region): 
    region_data =[x for x in data if x[1] == region]
    result1 = region_count(region_data, region)
    result2 = region_total(region_data, region)
    result3 = region_average(region_data, region)
    result4 = top_sale(region_data)
    print(f"==={region}===")
    print(f"Records: {result1}")
    print(f"Total Sale: ${result2:.2f}")
    print(f"Average Sale: ${result3:.2f}")
    print(f"Top Sale: {result4[0]} - ${result4[2]:.2f}")
```

```python
region_summary(sales_data, 'East')
```

```python
region_summary(sales_data, 'South')
```

```python
region_summary(sales_data, 'West')
```
