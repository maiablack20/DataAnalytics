## Lab 2. Working with datetime Functions 
## This exercise demonstrates the usage of datettime functions

# Today

```python
import datetime
```

```python
today = datetime.datetime.now()
```

```python
print(f"Today is", today.strftime("%A, %B %d, %Y"))
```

# My Birthday (The Best Day)

```python
my_birthday = datetime.date(1999, 3, 20)
```

```python
print(f"My birthday is", my_birthday.strftime("%m/%d/%Y"))
```

# Next 90 Days

```python
ninety_d = today + datetime.timedelta(days=90)
```

```python
print(f"90 days from today is", ninety_d.strftime("%B %d, %Y"))
```

# Time for Dinner

```python
dinner_time = datetime.time(20, 30, 00)
```

```python
print(f"Let's meet for dinner at" , dinner_time.strftime("%I:%M %p"))
```
