## In this lab, I will be working with functions for taxes within both Jupyter Notebook and VS Code as well. 

```python
def get_soc_sec_tax(gross_pay):
    tax = gross_pay * 0.062
    return tax
```

```python
def get_medicare_tax(gross_pay): 
    tax = gross_pay * 0.0145
    return tax
```

```python
def get_federal_tax (gross_pay, withholding_code):
    if withholding_code == 0:
        federal_tax = gross_pay * 0.23
    elif withholding_code == 1:
        federal_tax = gross_pay * 0.21
    elif withholding_code == 2: 
        federal_tax = gross_pay * 0.195
    elif withholding_code == 3: 
        federal_tax = gross_pay * 0.185
    elif withholding_code >= 4: 
        federal_tax = gross_pay * (0.18 - 0.005 * (withholding_code - 4))
    return federal_tax
```

```python
# person 1:
print("Social Security Tax: ", get_soc_sec_tax(750), "Medicare Tax: ",
get_medicare_tax(750), "Federal Tax: ",
get_federal_tax(750, 0))
```
