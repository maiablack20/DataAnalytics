```python
def favorite_things(name, movie, music):
    print(f"Name: {name}")
    print(f"Movie: {movie}")
    print(f"Music: {music}")
```

```python
favorite_things("Maia", "Coraline", "Monaleo")
```

```python
def why_im_here(purpose):
    print(f"Purpose: {purpose}")
```

```python
why_im_here("I am here because I want to be able to work in a career field where I am not only paid extremely well for my skill set, but to break free of customer service roles for good. I want something more myself than customer service or even retail. I deserve it.")
```

```python
def favorite_place(place, why):
    print(f"One of my favorite places is {place} because {why}.")
```

```python
favorite_place("Sweet Hut", "I love their Thai Iced Teas and mango mousse. It has a special place in my heart and I love going there.")
```
