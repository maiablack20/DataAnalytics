## Lab 1. Working with names - parsing strings

```python
test_name = "Miracle Evans"
```

```python
def parse_and_display(name):
    for i in range(len(name)):
        if name[i] == ' ':
            first = name[:i]
            last = name[i+1:]
            print(f"Full name: {name}")
            print(f"First name: {first}")
            print(f"Last name: {last}")
```

```python
parse_and_display(test_name)
```

```python
test_list = ['Grace Flores', 'JB Oinonen', 'Ken Penn']
```

```python
parse_and_display(test_list[0]) # output for Grace Flores
parse_and_display(test_list[1]) # output for JB Oinonen
parse_and_display(test_list[2]) # output for Ken Penn
```

```python
test_name = 'Maia Black' 
parse_and_display(test_name) # output for Maia Black
```

```python
def parse_and_display2(name2): 
    for i in range(len(name2)):
        if name2[i] == ' ':
            first = name2[:i]
            break
    for i2 in range(i + 1, len(name2)): 
        if name2[i2] == ' ':
            middle = name2[i  + 1:i2]
            last = name2[i2 + 1:] 
            break
    print(f"Full name: {name2}")
    print(f"First name: {first}")
    print(f"Middle name: {middle}")
    print(f"Last name: {last}")
```

```python
test_name2 = 'Megan Thee Stallion'
```

```python
parse_and_display2(test_name2)
```
