## Lab 2. Parsing strings - part codes

```python
def get_supplier_code(code):
    for i in range(len(code)):
        if code[i] == ':':
            supplier_code = code[:i]
            break
    return supplier_code
```

```python
code = get_supplier_code('ACME:123-L')
code = get_supplier_code('DI:12345-M')
code = get_supplier_code('ACE:1-12')
```

```python
print(get_supplier_code('ACME:123-L'))
print(get_supplier_code('DI:12345-M'))
print(get_supplier_code('ACE:1-12'))
```

```python
def get_product_num(num): 
    for i in range(len(num)):
        if num[i] == ':':
            colon_pos = i
            break
    for i in range((colon_pos + 1), len(num)):
        if num[i] == '-':
            dash_pos = i 
            product_num = num[colon_pos + 1: i]
            break
    return product_num
```

```python
num = get_product_num('ACME:123-L')
num = get_product_num('DI:12345-M')
num = get_product_num('ACE:1-12')
```

```python
print(get_product_num('ACME:123-L'))
print(get_product_num('DI:12345-M'))
print(get_product_num('ACE:1-12'))
```

```python
def get_size(size):
    for i in range(len(size)):
        if size[i] == '-':
            size_code = size[i + 1:]
            break 
    return size_code
```

```python
size = get_size('ACME:123-L')
size = get_size('DI:12345-M')
size = get_size('ACE:1-12')
```

```python
print(get_size('ACME:123-L'))
print(get_size('DI:12345-M'))
print(get_size('ACE:1-12'))
```

```python
part_code = 'ACME:123-L'
print(f"For {part_code}, the supplier is {get_supplier_code(part_code)}. Part number {get_product_num(part_code)} is available in size {get_size(part_code)}")
```
