## Math with Integers and Floats

### Understanding the difference between `/` and `//` (and how this impacts datatypes)

Write an expression to illustrate the answer to each question below. Make sure to run the cell to check your results.

```python
# What is the result of 2 times 2?

a = 2
b = 2
c = a * b
print(c)

    # The result of 2 times 2 is 4. 
```

```python
# What is the result of 4 divided by 2, using regular division?

a = 4 
b = 2 
c = 4/2 
print(c) 

    # The result of 4 divided by 2 is 2. 
```

```python
# What is the result of 4 divided by 2, using floor division?

a = 4 
b = 2
c = 4//2 
print(c)

    # The result of 4 divided by 2 is 2 after using floor division. The result from what I observed is the same as if one was using regular division for this, which means that floor division is much better when you are dividing numerals together that produce a number with a decimal.
```

```python
# Ok, how about 2 times 2.5?

a = 2 
b = 2.5 
c = 2 * 2.5 
print(c)

    # The result of 2 times 2.5 is 5. 
```

```python
# What's 5 divided by 2 using regular division?

a = 5 
b = 2
c = a/b 
print(c)

    # The result of 5 divided by 2 is 2.5 
```

```python
# What's 5 divided by 2 using floor division?

a = 5 
b = 2
c = 5//2
print(c)

    # The result of 5 divided by 2 using floor division is 2. 
```

```python
# When might you want to use floor division instead of normal division? (Write your answer as a code comment below.)

    # When you would use floor division instead of normal division is when you're dividing numbers like "6" by "4", which would usually give you a number with a decimal, which it rounds to the nearest whole number. 
```
